```python
import random
import time
from kafka import KafkaProducer
import json
import threading

def generate_random_location():
    latitude = random.uniform(-90, 90)
    longitude = random.uniform(-180, 180)
    return {"latitude": latitude, "longitude": longitude}

def send_truck_location(producer, topic, partition, truck_id):
    while True:
        location = generate_random_location()
        location_data = {"truck_id": truck_id, "location": location}
        producer.send(topic, value=location_data, partition=partition)
        print(f"Truck {truck_id} sent location: {location}")
        time.sleep(1)

def main():
    num_trucks = 3  # Change this to the number of trucks you want number of trucks should be <= number of partitions
    topic = 'TruckLocations'

    producer = KafkaProducer(
        bootstrap_servers=['localhost:29092'],
        value_serializer=lambda v: json.dumps(v).encode('utf-8')
    )

    threads = []
    for truck_id in range(num_trucks):
        partition = truck_id  # Assign each truck its own partition
        thread = threading.Thread(target=send_truck_location, args=(producer, topic, partition, truck_id))
        threads.append(thread)
        thread.start()

    # Wait for all threads to finish (this will actually never happen because of the infinite loop)
    for thread in threads:
        thread.join()

if __name__ == "__main__":
    main()
```
